```python
# app.py
import streamlit as st
import pandas as pd
import os
import re

# READING IN THE LATEST FILE
# ----------------------------------------------------------------------------------------
# Directory containing the files
folder = "/Users/noahwebb/Python/scraping_env2/pokeScrape/serbii_data"

# List all files in the folder
all_files = os.listdir(folder)
print(all_files)

# Function to extract date from filename
def extract_date(filename):
    match = re.match(r"serbii_data_(\d{8})_(\d{6})\.csv", filename)
    return int(match.group(1)) if match else None

# Collect valid files with extracted dates
file_data = []
for f in all_files:
    date = extract_date(f)
    if date:
        file_data.append({'file': os.path.join(folder, f), 'date': date})

file_data
```

```python

# Convert to DataFrame and sort
df_files = pd.DataFrame(file_data)
df_files_sorted = df_files.sort_values(by='date', ascending=False)

# Load the latest file
if not df_files_sorted.empty:
    latest_file = df_files_sorted.iloc[0]['file']
    df = pd.read_csv(latest_file)
    print(f"Loaded: {latest_file}")
else:
    raise FileNotFoundError("No matching Pokedex_YYYYMMDD.csv files found.")

df = pd.read_csv(latest_file)  # Or build from scraped data
```

```python
# ----------------------------------------------------------------------------------------

# Extract unique types (split and flatten the list)
all_types = set()
for types in df['Type']:
    for t in types.split('|'):
        all_types.add(t.strip())
all_types = sorted(all_types)

# Create multiselect filter
selected_types = st.multiselect("Filter by Type:", all_types)

# Apply filter
if selected_types:
    # Keep rows where any selected type appears in the 'Type' column
    df_filtered = df[df['Type'].apply(lambda x: any(t in x for t in selected_types))]
else:
    df_filtered = df

# Show filtered table
st.dataframe(df_filtered)

# Text input filter for name
name_filter = st.text_input("Search by Name:")

# Apply name filter
if name_filter:
    df_filtered = df_filtered[df_filtered['Name'].str.contains(name_filter, case=False)]

# ----------------------------------------------------------------------------------------

st.set_page_config(layout="centered")

st.title("Pokémon Pokedex")

st.write("This app displays data from a CSV file and lets you explore it interactively.")
st.write("")  # This adds a blank line


# st.markdown("""
# Welcome to the **CSV Table Viewer** app!

# Here you can:
# - Upload your CSV file
# - View data tables
# - Filter and sort data easily
# """)

# Override Streamlit padding just for this section
# st.markdown(
#     """
#     <style>
#     /* Remove horizontal padding from all main containers */
#     .block-container {
#         padding-left: 0rem !important;
#         padding-right: 0rem !important;
#         max-width: 100% !important;
#         margin: 50 !important;
#     }

#     /* Table container styling */
#     .wide-table-container {
#         width: 100%;
#         overflow-x: auto;
#         border: 1px solid #ddd;
#         padding: 10px;
#         margin: 50;
#     }

#     table {
#         border-collapse: collapse;
#         width: 100%;
#     }

#     th, td {
#         padding: 8px 12px;
#         border: 1px solid #ccc;
#         white-space: nowrap;
#         text-align: left;
#     }
#     </style>
#     """,
#     unsafe_allow_html=True,
# )


# # # Wrap the table in a div with class 'table-container'
# table_html = df.to_html(classes='wide-table-container')

# # # Display table with scroll container
# st.markdown(f"<div class='wide-table-container'>{table_html}</div>", unsafe_allow_html=True)







# default dataframe
st.dataframe(df)

```
